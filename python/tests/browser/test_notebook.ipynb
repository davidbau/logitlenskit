# Widget Display Test
This notebook tests that multiple widgets display correctly in a loop.

```python
import sys
sys.path.insert(0, '../src')

import torch
from logitlenskit.display import show_logit_lens, display_logit_lens
```

```python
# Create test data
def make_data(input_tokens, vocab_map, model_name):
    n_pos = len(input_tokens)
    n_layers = 3
    k = 2
    vocab_ids = list(vocab_map.keys())
    
    topk = torch.zeros(n_layers, n_pos, k, dtype=torch.int32)
    tracked = []
    probs = []
    
    for pos in range(n_pos):
        for li in range(n_layers):
            topk[li, pos] = torch.tensor(vocab_ids[:k], dtype=torch.int32)
        tracked.append(torch.tensor(vocab_ids[:k], dtype=torch.int32))
        probs.append(torch.rand(n_layers, k))
    
    return {
        'model': model_name,
        'input': input_tokens,
        'layers': list(range(n_layers)),
        'topk': topk,
        'tracked': tracked,
        'probs': probs,
        'vocab': vocab_map,
    }

data1 = make_data(['The', ' capital', ' of', ' France'], {1: 'Paris', 2: 'London'}, 'model-1')
data2 = make_data(['Hello', ' world'], {10: 'Hi', 11: 'Hey'}, 'model-2')
data3 = make_data(['1', ' +', ' 1', ' ='], {20: '2', 21: '3'}, 'model-3')
```

```python
# Test 1: Display widgets in a loop
datasets = [
    (data1, 'Widget 1: France'),
    (data2, 'Widget 2: Hello'),
    (data3, 'Widget 3: Math'),
]

for data, title in datasets:
    display_logit_lens(data, title=title)
```

## Expected Results

- **Widget 1** should show 4 input tokens: 'The', ' capital', ' of', ' France'
- **Widget 2** should show 2 input tokens: 'Hello', ' world'
- **Widget 3** should show 4 input tokens: '1', ' +', ' 1', ' ='

Each widget should have its own title and data.
